```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report
```

```python
spam = pd.read_csv('spam.csv', encoding='latin-1')
spam.columns = ['type', 'text', '', '', '']
spam['type'] = spam['type'].map({'ham': 0, 'spam': 1})
spam.drop(['', '', '', ''], axis=1, inplace=True)
spam.head()
```

```python
null_total = spam.isnull().sum()
print(null_total)
```

```python
spam.shape
```

```python
spam.head(100)
```

**TRAIN DATA**

```python
X = spam['text']
y = spam['type']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=3)
```

```python
vectorizer = TfidfVectorizer()
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)
```

**LOGISTIC REGRESSION**

```python
LR = LogisticRegression()
LR.fit(X_train_tfidf, y_train)
y_pred_logistic = LR.predict(X_test_tfidf)
```

**SVM**

```python
SVM = SVC()
SVM.fit(X_train_tfidf, y_train)
y_pred_svm = SVM.predict(X_test_tfidf)
```

**REPORT**

```python
#Logistic Regression
lr_report = classification_report(y_test, y_pred_logistic)
print(f"Logistic Regression Classification Report: ")
print(lr_report)

accuracy_logistic = accuracy_score(y_test, y_pred_logistic)
print(f"Logistic Regression Accuracy: {accuracy_logistic} \n")

#SVM
svm_report = classification_report(y_test, y_pred_svm)
print(f"SVM Classification Report: ")
print(svm_report)

accuracy_svm = accuracy_score(y_test, y_pred_svm)
print(f"SVM Accuracy: {accuracy_svm}")
```

**Difference of Performance**

*Overall Accuracy*
*   SVM slightly outperforms Logistic Regression (98.65% vs 97.57%)

*Spam Detection*
*   SVM has better recall (0.90 vs 0.83) → which means it catches more spam
*   SVM has better F1-score for spam (0.95 vs 0.90) → which means SVM has a better balance between: Precision (correctly identifying actual spam) and
Recall (catching as many spam emails as possible)
